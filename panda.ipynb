```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Age': [25, 30, 28, 35, 40, 29, 32, 27],
    'Gender': ['F', 'M', 'F', 'M', 'M', 'F', 'M', 'F'],
    'Department': ['HR', 'Sales', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'HR'],
    'Experience': [2, 5, 4, 8, 10, 3, 6, 2],
    'Salary': [22000, 35000, 40000, 55000, 60000, 38000, 48000, 21000],
    'Sales': [150000, 320000, 280000, 450000, 200000, 300000, 410000, 160000]
}
```

```python
# Import Pandas
import pandas as pd

# Convert the data dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Last 5 rows:")
display(df.tail())
```

```python
print("Shape of the dataset:")
print(df.shape)
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
avg_salary = df.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
print(avg_salary)
```

```python
total_sales = df.groupby('Department')['Sales'].sum()
print("\nTotal Sales by Department:")
print(total_sales)
```

```python
highest_sales_employee = df.loc[df['Sales'].idxmax()]
print("Employee with the highest sales:")
print(highest_sales_employee)

```

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)


IQR = Q3 - Q1


lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)


outliers = df[(df['Salary'] < lower_limit) | 
              (df['Salary'] > upper_limit)]

print("\nSalary Outliers:")
print(outliers)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Age'], bins=6, kde=True)

plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.show()
```

```python

dept_salary = df.groupby('Department')['Salary'].mean()


plt.figure(figsize=(8, 5))

dept_salary.plot(kind='bar')

plt.title('Department-wise Average Salary')
plt.xlabel('Department')
plt.ylabel('Average Salary')

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Experience',
    y='Salary',
    s=100
)

plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')

plt.show()
```

```python
correlation = df['Salary'].corr(df['Sales'])

print("Correlation between Salary and Sales:")
print(correlation)
```

```python
correlation_matrix = df[['Age', 'Experience', 'Salary', 'Sales']].corr()

# Plot the correlation matrix
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')

plt.show()
```

### Key Insights

1. The Sales department has the highest average salary of ₹46,000 and

2. Employee 104 has the highest individual sales of ₹4,50,000.

3. There is a strong positive relationship between experience and salary,
